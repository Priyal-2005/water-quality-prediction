```python
# Cell Description: Load the Ireland dataset and display its dimensions and column names.

import pandas as pd
import numpy as np

# Load the Ireland dataset from the raw data directory
df_ireland = pd.read_csv("../data/raw/Ireland_dataset.csv", low_memory=False)

# Display the total number of rows and columns
print(f"Total Rows: {len(df_ireland):,}")
print(f"Total Columns: {len(df_ireland.columns)}")

# Display all column names
print("\nAll Column Names:")
print(list(df_ireland.columns))
```

```python
# Cell Description: Identify location and date columns and inspect their unique values and sample entries.

print("--- CHECKING QUESTION 1: Site / Station ID ---")

# Check available location-related columns and count unique locations
location_columns = [
    'Country', 'Area', 'Waterbody Type',
    'Station_ID', 'Site_ID', 'MonitoringLocationIdentifier'
]

for col in location_columns:
    if col in df_ireland.columns:
        print(
            f"Column '{col}' exists! "
            f"Number of unique locations: {df_ireland[col].nunique()}"
        )

print("\n--- CHECKING QUESTION 2: Date ---")

# Identify available date-related columns and display sample values
date_columns = ['Date', 'sample_date', 'Year', 'date']

for col in date_columns:
    if col in df_ireland.columns:
        print(f"Column '{col}' exists!")
        print(f"Sample dates: {df_ireland[col].dropna().head(3).tolist()}")
```

```python
# Cell Description: Analyze missing data across columns and identify existing WQI or water-quality target columns.

print("--- CHECKING QUESTION 3: Parameters & Missing Data ---")

# Calculate the percentage of missing values in each column
missing_pct = (df_ireland.isnull().mean() * 100).round(2)

print("Missing Percentage per Column (%):")
print(missing_pct)

print("\n--- CHECKING QUESTION 4: Existing WQI / Target Column ---")

# Identify columns potentially related to WQI, water quality, or classification targets
target_keywords = ['wqi', 'ccme', 'class', 'status', 'value']

target_cols = [
    col for col in df_ireland.columns
    if any(keyword in col.lower() for keyword in target_keywords)
]

print(f"Target/WQI Columns Found: {target_cols}")
```

```python
# Cell Description: Summarize the Ireland dataset's continuous CCME values and categorical WQI class distribution.

# In notebooks/01_dataset_inspection.ipynb

print("--- TARGET SUMMARY (Ireland Dataset) ---")

# 1. Summarize the continuous CCME values
print("\n1. Continuous Target (CCME_Values) Summary:")
print(df_ireland['CCME_Values'].describe())

# 2. Calculate the percentage distribution of CCME_WQI classes
print("\n2. Categorical Target (CCME_WQI) Class Distribution:")
print(df_ireland['CCME_WQI'].value_counts(normalize=True).mul(100).round(2))
```

```python
# Verification of Target & Feature Separation for Priyal
y_reg = df_ireland["CCME_Values"]
y_clf = df_ireland["CCME_WQI"]

# Parameters intentionally excluded to prevent CCME formula leakage
leakage_columns = [
    "Biochemical Oxygen Demand (mg/l)",
    "Dissolved Oxygen (mg/l)",
    "Nitrate (mg/l)"
]

# Primary model feature candidates
feature_columns = [
    "pH (ph units)",
    "Temperature (cel)",
    "Ammonia (mg/l)",
    "Orthophosphate (mg/l)",
    "Nitrogen (mg/l)",
    "Waterbody Type"
]

X_preview = df_ireland[feature_columns]

print("=== ZERO-LEAKAGE PARTITION CONFIRMATION ===")
print(f"Target y (Regression) shape:    {y_reg.shape}")
print(f"Target y (Classification) shape:{y_clf.shape}")
print(f"Feature matrix X shape:          {X_preview.shape}")
print(f"\nFeatures in X: {list(X_preview.columns)}")
print(f"Excluded from X (Leakage Guard): {leakage_columns}")

# Ensure complete set separation
assert len(set(X_preview.columns).intersection(set(leakage_columns))) == 0
assert "CCME_Values" not in X_preview.columns
assert "CCME_WQI" not in X_preview.columns
print("\n--> PASSED: 0% Direct Mathematical Formula Leakage!")
```
